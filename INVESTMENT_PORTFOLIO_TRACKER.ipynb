```python
import pandas as pd
from openpyxl import Workbook
from openpyxl.worksheet.table import Table, TableStyleInfo
from openpyxl.styles import Font, Alignment, PatternFill

# Portfolio tracking template data - no nulls
rows = [
    [1, 'AAPL', 'Apple Inc.', 'Technology', 50, 180.25, 9000.00, 195.10, 750.00, 8.33, 'Long-term', 'Active'],
    [2, 'MSFT', 'Microsoft Corp.', 'Technology', 30, 410.00, 12300.00, 425.75, 472.50, 3.84, 'Dividend Growth', 'Active'],
    [3, 'VOO', 'Vanguard S&P 500 ETF', 'ETF', 20, 430.00, 8600.00, 450.20, 404.00, 4.70, 'Core Holding', 'Active'],
    [4, 'JNJ', 'Johnson & Johnson', 'Healthcare', 40, 160.00, 6400.00, 155.80, -168.00, -2.63, 'Defensive', 'Active'],
    [5, 'TSLA', 'Tesla Inc.', 'Automotive', 10, 220.00, 2200.00, 205.50, -145.00, -6.59, 'Speculative', 'Active'],
]

cols = [
    'ID', 'Ticker', 'Asset Name', 'Sector', 'Quantity', 'Buy Price', 'Cost Basis',
    'Current Price', 'Unrealized P/L', 'Return %', 'Investment Strategy', 'Status'
]

df = pd.DataFrame(rows, columns=cols)

# Save CSV
csv_name = 'investment_portfolio_tracking.csv'
df.to_csv(csv_name, index=False)

# Create formatted Excel workbook
wb = Workbook()
ws = wb.active
ws.title = 'Portfolio'

# Title
ws['A1'] = 'Investment Portfolio Tracker'
ws['A1'].font = Font(bold=True, size=14)
ws['A1'].alignment = Alignment(horizontal='center')

# Headers and data start at row 3
for c, col_name in enumerate(cols, start=1):
    cell = ws.cell(row=3, column=c, value=col_name)
    cell.font = Font(bold=True, color='FFFFFF')
    cell.fill = PatternFill('solid', fgColor='1F4E78')
    cell.alignment = Alignment(horizontal='center')

for r_idx, row in enumerate(rows, start=4):
    for c_idx, value in enumerate(row, start=1):
        ws.cell(row=r_idx, column=c_idx, value=value)

# Add table
end_row = 3 + len(rows)
end_col = len(cols)
tab = Table(displayName='PortfolioTable', ref=f'A3:L{end_row}')
style = TableStyleInfo(name='TableStyleMedium2', showFirstColumn=False, showLastColumn=False, showRowStripes=True, showColumnStripes=False)
tab.tableStyleInfo = style
ws.add_table(tab)

# Widths
widths = [8, 12, 24, 16, 10, 12, 12, 14, 14, 10, 20, 12]
```

```python
for i, width in enumerate(widths, start=1):
    ws.column_dimensions[chr(64 + i)].width = width

# Save workbook
excel_name = 'investment_portfolio_tracking.xlsx'
wb.save(excel_name)

print(f"CSV saved as '{csv_name}'")
print(f"Excel workbook saved as '{excel_name}'")
```

```python
sector_performance = df.groupby('Sector').agg(
    total_cost_basis=('Cost Basis', 'sum'),
    total_unrealized_pl=('Unrealized P/L', 'sum'),
    average_return_percent=('Return %', 'mean')
).reset_index()

display(sector_performance)
```
